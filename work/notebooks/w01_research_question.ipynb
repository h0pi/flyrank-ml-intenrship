## 2. The question: decision, action, and cost of a wrong call

### Search question
Which existing content pages should an SEO or content team prioritize for review because they show signals consistent with declining search performance while still having meaningful visibility?

### Unit of analysis
One content page.

### Output
A ranked page-level review queue with a priority score. Higher-ranked pages should be reviewed earlier.

### Decision
A content or SEO manager decides which pages should be assigned for review first.

### Action
For each high-priority page, an editor or SEO specialist can inspect search intent, content freshness, title and meta description, internal linking, position, CTR, and potential technical issues. They can then decide whether to update, consolidate, redirect, or leave the page unchanged.

### Cost of a wrong recommendation
A false positive costs editor and SEO time on a page that was not an important opportunity. A false negative means a genuinely declining page may not be reviewed in time and could continue losing organic visibility or traffic.

### Why data or ML can help
A team cannot manually inspect every page consistently at scale. Multiple observable signals can be combined to rank pages more systematically than a single fixed rule. ML is useful only if it provides a more useful, honestly validated ranking than a transparent baseline rule.

### Task type and success metric
This is a ranking/scoring task because the decision is “which pages should be reviewed first?”. An initial metric is Precision@50: among the 50 pages ranked highest, what proportion truly have a later observed decline. The final target and validation will use time-aware warehouse data.

## 3. Quick look at the data

The starter dataset contains 30,000 pseudonymized content items from 32 clients. I will use the following code to measure how many pages are currently marked as declining, how many still have meaningful search visibility, and how long pages have typically gone without an update.

These numbers do not prove that any feature causes decline. They show that the dataset contains enough page-level variation to make prioritization a useful decision-support problem.

```python
import os
import sys
import subprocess
import pandas as pd

repo_url = "https://github.com/h0pi/flyrank-ml-intenrship.git"
repo_dir = "/content/flyrank-ml-intenrship"

if not os.path.exists(repo_dir):
    subprocess.run(
        ["git", "clone", "--depth", "1", repo_url, repo_dir],
        check=True
    )

os.chdir(repo_dir)

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = df["trend_direction"].str.lower().eq("down").mean()
visible_pages = (df["impressions_90d"] >= 500).sum()
median_days_since_update = df["days_since_last_update"].median()

print(f"Total pages: {len(df):,}")
print(f"Declining-page rate in the starter label: {declining_rate:.1%}")
print(f"Pages with at least 500 impressions in 90 days: {visible_pages:,}")
print(f"Median days since last update: {median_days_since_update:.0f}")
```

The starter dataset contains 30,000 pseudonymized content pages. In the starter label, 54.2% of pages are marked as declining, which shows that prioritization cannot be based only on a few obvious cases.

There are 16,726 pages with at least 500 impressions in the trailing 90 days. These pages still have meaningful search visibility, so choosing which ones to review first can affect where editorial effort is spent.

The median page was updated 20 days ago. This suggests that time since the last update alone is not enough to decide which pages should be reviewed; a useful ranking should combine multiple signals.

These are observed descriptive numbers from the starter dataset. They support the value of a page-level prioritization workflow, but they do not show that any individual feature causes search decline.
